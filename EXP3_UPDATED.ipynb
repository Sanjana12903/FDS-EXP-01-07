```python
import zipfile

zip_path = "playground-series-s4e11.zip"
extract_path = "playground-series-s4e11"

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset extracted successfully!")
```

```python
import pandas as pd
loc=extract_path + '/' + "train.csv"
df = pd.read_csv(loc)
df.head()
```
