```python
import zipfile

zip_path = "playground-series-s4e3.zip"
extract_path = "playground-series-s4e3"

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset extracted successfully!")
```

```python
import pandas as pd
loc=extract_path + '/' + "train.csv"
df = pd.read_csv(loc)
df.head()
```

```python
df.corr(method="pearson")
```

```python
df.corr(method="pearson")["Minimum_of_Luminosity"]
```

```python
# correlation between "Dirtiness" vs all
df.corr(method="pearson")["Dirtiness"]
```

```python
# pearson correlation coefficient between "Dirtiness" vs "Minimum_of_Luminosity"
df.corr(method="pearson")["Dirtiness"]["Minimum_of_Luminosity"]
```
