```python
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings("ignore")
```

```python

loc = "playground-series-s4e3" + "/" + "train.csv"
df = pd.read_csv(loc)
df
```

```python

sample_size = 15
sample = np.random.choice(df["X_Minimum"], size=sample_size)
sample
```

```python
sample_2 = np.random.choice(df["X_Minimum"], size=sample_size)
sample_2
```

```python

print(f"Standard Deviation of 1st sample from the column 'X_Minimum' of df is {sample.std()}.")
print(f"Standard Deviation of 2nd sample from the column 'X_Minimum' of df is {sample_2.std()}.")
print(f"Standard Deviation of the population from the column 'X_Minimum' of df is {df['X_Minimum'].std()}.")
```

```python

print(f"Median of 1st sample from the column 'X_Minimum' of df is {np.median(sample)}.")
print(f"Median of 2nd sample from the column 'X_Minimum' of df is {np.median(sample_2)}.")
print(f"Median of the population from the column 'X_Minimum' of df is {np.median(df['X_Minimum'])}.")
```

```python
sample_mean = []
for i in range(1000):
    sample = np.random.choice(df["X_Minimum"], size=50)
    sample_mean.append(np.mean(sample))
```

```python
import matplotlib.pyplot as plt
plt.hist(sample_mean)
plt.xlabel("Sampling Distribution")
plt.ylabel("Freqency")
plt.show()
```
