```python
import pandas as pd
import numpy as np

loc = "playground-series-s4e11" + "/" + "train.csv"

# Load the dataset into a pandas DataFrame
df = pd.read_csv(loc)

# Display the first few rows to verify loading
df.head(15)
```

```python
# Simple Random Sampling
#Randomly select 15 names from the Name column-repitation allowed.
print(np.random.choice(df["Name"], 25, replace=True))
```

```python
#no repitation allowed
print(np.random.choice(df["Name"],10, replace=False))
```

```python
#Systematic Random Sampling
#Instead of randomly picking every person, you pick every fixed interval
interval = len(df) // 50
print(df["Name"][::interval][:50].tolist())
```

```python
#Stratified Random Sampling
#divide the population into groups, then randomly select people from each group.
sample_df = df.groupby("Gender").sample(n=500)
print(sample_df["Gender"].value_counts())
```

```python
# Cluster Sampling
cluster = df["Working Professional or Student"].unique()
cluster_df = df[df["Working Professional or Student"] == cluster[0]]
print(cluster_df.head(10))
```

```python
# Cluster Sampling

cluster = df["Working Professional or Student"].unique()

# Randomly select one cluster
random_cluster = np.random.choice(cluster)

# Select all rows belonging to that cluster
cluster_df = df[df["Working Professional or Student"] == random_cluster]

print("Selected Cluster:", random_cluster)
print(cluster_df.head())
```
