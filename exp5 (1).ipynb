```python
import zipfile

zip_path = "playground-series-s4e3.zip"
extract_path = "playground-series-s4e3"

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset extracted successfully!")
```

```python
import pandas as pd
loc=extract_path + '/' + "train.csv"
df = pd.read_csv(loc)
df.head()
```

```python

def negligible_category(df):
    # 1. Compute the Pearson Correlation matrix
    corr_df = df.corr(method="pearson")
    cols = list(corr_df.columns)
    
    # Lists to store the filtered pairs and values
    pairs_list = []
    corr_value = []
    
    # 2. Loop through the lower triangle of the correlation matrix to avoid duplicates
    for i in range(1, len(cols), 1):
        for j in range(0, i, 1):
            val = corr_df[cols[i]][cols[j]]
            # Check if absolute correlation is between 0 and 0.2
            if (abs(val) >= 0) and (abs(val) < 0.2):
                pairs_list.append(f"{cols[i]} vs {cols[j]}")
                corr_value.append(val)
                
    # 3. Create the output summary DataFrame
    negligible_df = pd.DataFrame()
    negligible_df["Entity"] = pairs_list
    negligible_df["Correlation Coefficient"] = corr_value
    
    return negligible_df

# 4. Call the function on your loaded DataFrame
# Make sure your 'df' is loaded (e.g., df = pd.read_csv("your_file.csv"))
result_df = negligible_category(df)
result_df
```

```python
def weak_category(df):
    # 1. Compute the Pearson Correlation matrix
    corr_df = df.corr(method="pearson")
    cols = list(corr_df.columns)
    
    # Storage lists for filtered feature combinations and coefficients
    pairs_list = []
    corr_value = []
    
    # 2. Iterate across unique column pairings (lower triangular half)
    for i in range(1, len(cols), 1):
        for j in range(0, i, 1):
            val = corr_df[cols[i]][cols[j]]
            # Filter condition: absolute correlation between 0.2 and 0.4
            if (abs(val) >= 0.2) and (abs(val) < 0.4):
                pairs_list.append(f"{cols[i]} vs {cols[j]}")
                corr_value.append(val)
                
    # 3. Structuralize outcomes into a result DataFrame
    weak_df = pd.DataFrame()
    weak_df["Entity"] = pairs_list
    weak_df["Correlation Coefficient"] = corr_value
    
    return weak_df

# 4. Execute the function on your dataset
result_weak_df = weak_category(df)
result_weak_df
```

```python

def strong_category(df):
    corr_df = df.corr(method="pearson")
    cols = list(corr_df.columns)
    
    # Renamed variable to avoid conflict with the function name
    strong_pairs, corr_value = [], []
    
    for i in range(1, len(cols)):
        for j in range(0, i):
            val = corr_df[cols[i]][cols[j]]
            if 0.6 <= abs(val) < 0.8:
                strong_pairs.append(f"{cols[i]} vs {cols[j]}")
                corr_value.append(val)
                
    strong_df = pd.DataFrame()
    strong_df["Entity"] = strong_pairs
    strong_df["Correlation Coefficient"] = corr_value
    
    return strong_df
    strong_category(df)
```
