```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")

# Load and Inspect Data
df = pd.read_csv("playground-series-s4e3/train.csv")

print("--- First 5 Rows ---")
print(df.head())

print("\n--- Data Info & Missing Values ---")
print(df.info())
print(df.isna().sum())

print("\n--- Summary Statistics ---")
print(df.describe())

# Visualizations
plt.figure(figsize=(18, 5))

# Histogram
plt.subplot(1, 3, 1)
plt.hist(df['Minimum_of_Luminosity'],
         bins=10, color='skyblue', edgecolor='black')
plt.title('Histogram of Minimum_of_Luminosity')
plt.xlabel('Values')
plt.ylabel('Frequency')

# Box Plot
plt.subplot(1, 3, 2)
sns.boxplot(x=df['Minimum_of_Luminosity'],
            color='lightcoral')
plt.title('Box Plot of Minimum_of_Luminosity')

# Distribution Plot
plt.subplot(1, 3, 3)
sns.histplot(df['Minimum_of_Luminosity'],
             kde=True, color='green', bins=10)
plt.title('Distribution Plot')

plt.tight_layout()
plt.show()

# Pair Plot
sns.pairplot(
    df[['Minimum_of_Luminosity',
        'SigmoidOfAreas', 'X_Maximum']]
)
plt.show()
```
