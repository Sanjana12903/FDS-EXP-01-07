```python
# 2. Systematic Random Sampling
sample_size = 50

# Calculate the interval step (k)
interval = len(df["Name"]) // sample_size
sample = []

# Select every kth entry from the Name column
for x in range(0, len(df["Name"]), interval):
    # If this throws a KeyError later, change df["Name"][x] to df["Name"].iloc[x]
    sample.append(df["Name"][x])

print(sample)
```

```python
# # Stratified Random Sampling
# Assuming 'Gender' is the column used for stratification
sample_size_of_each_strata = 500
category_column = df["Gender"]
category = category_column.value_counts().index
sample_df = pd.DataFrame()

for x in category:
    filtered_df = df[category_column == x]
    
    # Corrected 'randos.choice' to 'random.choice'
    random_index = np.random.choice(
        filtered_df.index, 
        size=sample_size_of_each_strata, 
        replace=False
    )
    
    # Corrected indent to keep it inside the loop block
    sample_df = pd.concat([sample_df, df.loc[random_index]])

# Reset the index of the final combined dataframe
sample_df.reset_index(inplace=True, drop=True)
```

```python
sample_df
```

```python
# Step 6: Verify the stratification counts
sample_df["Gender"].value_counts()
```

```python
# # Cluster Random Sampling
# Assuming 'Working Professional or Student' is the column used for Clustering
sample_size_of_each_strata = 1000
category_column = df["Working Professional or Student"]
category = category_column.value_counts().index

# Create separate DataFrames for each cluster group
cluster_1_df = df[category_column == category[0]]
cluster_2_df = df[category_column == category[1]]
```

```python
import pandas as pd
import numpy as np

loc = "playground-series-s4e11" + "/" + "train.csv"

# Load the dataset into a pandas DataFrame
df = pd.read_csv(loc)

# Display the first few rows to verify loading
df.head(15)
```

```python
# Simple Random Sampling
#Randomly select 15 names from the Name column-repitation allowed.
print(np.random.choice(df["Name"], 25, replace=True))
```

```python
#no repitation allowed
print(np.random.choice(df["Name"], 10, replace=False))
```

```python
#Systematic Random Sampling
#Instead of randomly picking every person, you pick every fixed interval
interval = len(df) // 50
print(df["Name"][::interval][:50].tolist())
```

```python
#Stratified Random Sampling
#divide the population into groups, then randomly select people from each group.
sample_df = df.groupby("Gender", group_keys=False).sample(n=500)
print(sample_df["Gender"].value_counts())
```

```python
# Cluster Sampling
cluster = df["Working Professional or Student"].unique()
cluster_df = df[df["Working Professional or Student"] == cluster[0]]
print(cluster_df)
```
